```python
import pandas as pd
import numpy as np

# Do not modify this cell.
# Raw HR employee dataset created for data cleaning practice.

employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103, 103, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115],
    "FirstName": ["Ainura", "Bakyt", "Gulnara", "Gulnara", "Nurlan", "Aigul", "Beksultan", "Aisuluu", "Ermek", "Nurgul", "Daniyar", "Aizhan", "Kubat", "Perizat", "Adilet"],
    "LastName": ["Asanova", "Kyzylov", "Mamatova", "Mamatova", "Abdyldaev", "Sultanova", "Toktogulov", "baisalova", "Kadyrov", "Jumabaeva ", "Turgunov", "Ryskulova", "Kadyraliev", "BAktybekova", "MOLDOKULOV"],
    "Department": ["IT", "HR", "it", "IT", "Finance", "HR", " IT ", "Finance", "IT", "HR", "it", "Financing", "Information Tecnology", "HR", "finance"],
    "Gender": ["F", "m", "F", "f", "M", "F", "M", "F", "M", "F", "M", "F", "M", "F", "m"],
    "Age": ["28", "35", "29", "29", "150", "31", "33", "27", "30", "38", "26", "34", "29", "32", "31"],
    "Salary": ["80000", "65000", np.nan, np.nan, "90000", "70000", "unknown", "62000", "85000", "58000", "72000", "66000", "78000", "88000", "61000"],
    "HireDate": ["2023-01-15", "2022-06-10", "2024-03-01", "2024-03-01", "2021-11-20", "2023-08-05", "2020-05-12", "2022-09-18", "2021-03-25", "2019-07-14", "2023-04-18", "2020-10-02", "2022-12-11", "2021-08-30", "2023-06-22"],
    "PerformanceScore": [4.2, 3.8, 4.5, 4.5, 4.0, 3.9, 4.1, 3.7, 4.3, 3.6, 4.4, 3.8, 4.2, "4.0", 3.9]
})

departments = pd.DataFrame({
    "DepartmentID": [1, 2, 3],
    "DepartmentName": ["IT", "HR", "Finance"],
    "Location": ["Bishkek", "Osh", "Bishkek"]
})
```

```python
df = employees.copy()
print("razmer:", df.shape)
print(df.dtypes)
print("propuski:")
print(df.isnull().sum())
print("salary, age, date:")
print(df[["Salary", "Age", "HireDate"]].head())
print("ocenki:", df["PerformanceScore"].unique())
print("nepravilniy vozrast:", df[df["Age"] == "150"]["EmployeeID"].values)
```

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
print("propuski:", df["Salary"].isnull().sum())
print("procent:", df["Salary"].isnull().mean() * 100)
print("mediana:", df["Salary"].median())

print("dropna vse:", df.dropna().shape)
print("dropna salary:", df.dropna(subset=["Salary"]).shape)
print("dropna thresh:", df.dropna(thresh=9).shape)

df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print("propuski posle:", df["Salary"].isnull().sum())
```

```python
df["Department"] = df["Department"].str.strip().str.lower()
df["Department"] = df["Department"].replace({
    "it": "IT",
    "information tecnology": "IT",
    "hr": "HR",
    "finance": "Finance",
    "financing": "Finance"
})
df["LastName"] = df["LastName"].str.strip().str.title()
df["Gender"] = df["Gender"].str.upper()
print("otdeli:", df["Department"].unique())
print("pol:", df["Gender"].unique())

df["Age"] = pd.to_numeric(df["Age"])
print("oshibka vozrasta:", df[df["Age"] > 100][["EmployeeID", "Age"]])
df.loc[df["Age"] > 100, "Age"] = np.nan
```

```python
print("polnie povtori:", df.duplicated().sum())
print("povtori ID:", df.duplicated(subset="EmployeeID").sum())
print("keep first:", df.drop_duplicates(subset="EmployeeID", keep="first").shape)
print("keep last:", df.drop_duplicates(subset="EmployeeID", keep="last").shape)
print("keep False:", df.drop_duplicates(subset="EmployeeID", keep=False).shape)
df = df.drop_duplicates(subset="EmployeeID", keep="first").copy()
print("posle:", df.shape)
```

```python
df["Age"] = df["Age"].astype("Int64")
df["Department"] = df["Department"].astype("category")
df["Gender"] = df["Gender"].astype("category")
df["PerformanceScore"] = pd.to_numeric(df["PerformanceScore"])
print(df.dtypes)
```

```python
df["HireDate"] = pd.to_datetime(df["HireDate"])
df["Year"] = df["HireDate"].dt.year
df["Month"] = df["HireDate"].dt.month
df["Weekday"] = df["HireDate"].dt.day_name()

# Priblizitelniy stazh v godah
df["YearsWorked"] = ((pd.Timestamp.today() - df["HireDate"]).dt.days / 365.25).round(1)
print(df[["EmployeeID", "Year", "Month", "Weekday", "YearsWorked"]])
```

```python
df = df.rename(columns={"Salary": "MonthlySalary", "HireDate": "EmploymentStartDate"})
itog = df.merge(departments, how="left", left_on="Department", right_on="DepartmentName")
print("razmer:", itog.shape)
print("propuski posle merge:")
print(itog[["DepartmentID", "Location"]].isnull().sum())
print(itog[["EmployeeID", "Department", "Location"]].head())
```

```python
print("razmer:", itog.shape)
print("tipi:")
print(itog.dtypes)
print("propuski:")
print(itog.isnull().sum())
print("povtori ID:", itog.duplicated(subset="EmployeeID").sum())
print("povtori strok:", itog.duplicated().sum())
print("propuski dat:", itog["EmploymentStartDate"].isnull().sum())
print("otdeli:", itog["Department"].unique())
```
