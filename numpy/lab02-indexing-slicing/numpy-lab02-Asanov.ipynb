```python
order_id = np.arange(2001, 2013)
region = np.array([ "Bishkek", "Osh", "Bishkek", "Karakol", "Osh", "Tokmok", "Bishkek", "Osh", "Karakol", "Tokmok", "Bishkek", "Osh" ])
quantity = np.array([2, 1, 5, 3, 2, 8, 1, 4, 2, 3, 6, 1])
unit_price = np.array([ 1500, 20000, 800, 5000, 1200, 700, 25000, 3000, 4500, 2000, 900, 18000 ])
discount = np.array([ 0.05, 0.10, 0.00, 0.15, 0.05, 0.20, 0.30, 0.10, 0.00, 0.05, 0.50, 0.10 ])
unit_cost = np.array([ 1000, 16000, 600, 4000, 900, 500, 18000, 2200, 3500, 1500, 800, 14000 ])
revenue = quantity * unit_price * (1 - discount) 
profit = revenue - quantity * unit_cost 


#proverka
assert (order_id.size == region.size == quantity.size == unit_price.size == discount.size == unit_cost.size)
assert np.all(quantity > 0)
assert np.all(unit_price >= 0)
assert np.all(unit_cost >= 0)
assert np.all((discount >= 0) & (discount <= 1))

porog = 10_000
region = "Bishkek"

vishe_poroga = revenue > porog
bolshe3 = quantity >= 3
vibranniy_region = region == region

vse_podhodit = vishe_poroga & bolshe3 & vibranniy_region

print("viruchka vishe poroga: ", order_id[vishe_poroga])
print("kolichestvo ne menshe 3: ", order_id[bolshe3])
print("Region: ", order_id[vibranniy_region])
print("vse usloviya pravilni: ", order_id[vse_podhodit])

assert np.array_equal(order_id[vse_podhodit], order_id[vishe_poroga & bolshe3 & vibranniy_region])
```

```python
skidka_bolshe20 = discount >= 0.20
viruchka_bolshe_10000 = revenue > 10_000
bolshe4str = quantity >= 4

kombinacia = skidka_bolshe20 & viruchka_bolshe_10000

print("skidka bolshe 20: ", skidka_bolshe20)
print("viruchka bolshe 10000: ", viruchka_bolshe_10000)
print("kolichestvo bolshe 4: ", bolshe4str)
print("kombinacia skidka bolshe 20 i viruchka bolshe 10000: ", order_id[kombinacia])
```

```python
minusi = profit < 0
orderminus = order_id[minusi]
kolichestvo = np.count_nonzero(minusi)

print("minusovie zakazy: ", orderminus)
print("ih kolichestvo: ", kolichestvo)
print("pribl posle minusov: ", profit[minusi])
print("chem bolshe viruchka, tem bolshe shans na minusovie zakazy")
```

```python
malo_skidki = discount < 0.10
mnogo_skidki = discount >= 0.20

print("zakazy s nizkoy skidkoy: ", order_id[malo_skidki])
print("pribl u nizkoy skidki: ", profit[malo_skidki])
print("srednyaya pribl: ", profit[malo_skidki].mean())

print("zakazy s bolshoy skidkoy: ", order_id[mnogo_skidki])
print("pribl u bolshoy skidki: ", profit[mnogo_skidki])
print("srednyaya prible: ", profit[mnogo_skidki].mean())
```

```python
mnogoskidkimalopribl = (discount >= 0.30) & (profit <= 1_000)
dorogoimalo = (unit_price >= 15_000) & (quantity <= 1)

podozrenie = mnogoskidkimalopribl | dorogoimalo

print("visokaya skidka i malo pribyli: ", order_id[mnogoskidkimalopribl])
print("dorogo no malo: ", order_id[dorogoimalo])
print("est podozreniye: ", order_id[podozrenie])
```
