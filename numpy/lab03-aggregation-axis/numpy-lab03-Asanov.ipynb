```python
import numpy as np 
sales = np.array([ [120, 125, 130, 128, 135, 140, 138, 145], [80, 82, 79, 85, 88, 90, 86, 92], [150, 148, 155, 160, 158, 165, 220, 170], [60, 65, 62, 58, 64, 63, 61, 66], [100, 105, 102, 108, 110, 112, 109, 115] ])
region_names = np.array([ "Bishkek", "Osh", "Karakol", "Tokmok", "Jalal-Abad" ])
```

```python
obshayaviruchka = np.sum(sales, axis=1)
srednie = np.mean(sales, axis=1)
mediana = np.median(sales, axis=1)

print("obshaya viruchka: ", obshayaviruchka)
print("srednie nedelniye prodaji: ", srednie)
print("mediana nedelnih: ", mediana)
print("lider obshey viruchki: ", region_names[np.argmax(obshayaviruchka)])
```

```python
razbros = np.std(sales, axis=1)
q1 = np.percentile(sales, 25, axis=1)
q3 = np.percentile(sales, 75, axis=1)
iqr = q3 - q1
koeff_razbrosa = razbros / srednie * 100

print("standartniy razbros: ", razbros)
print("mejkvartilniy razmah: ", iqr)
print("otnositelniy razbros v procentah: ", koeff_razbrosa)
print("samiy stabilniy region: ", region_names[np.argmin(koeff_razbrosa)])
print("samiy nestabilniy region: ", region_names[np.argmax(koeff_razbrosa)])
```

```python
niz_granica = q1 - 1.5 * iqr
verh_granica = q3 + 1.5 * iqr

vybros = (sales < niz_granica[:, np.newaxis]) | (sales > verh_granica[:, np.newaxis])

print("nizhnie granici: ", niz_granica)
print("verhnie granici: ", verh_granica)
print("maska vibrosov:")
print(vybros)
print("znacheniya vibrosov: ", sales[vybros])
print("region vibrosa: ", region_names[np.where(vybros)[0]])
```

```python
obshaya_po_magazinu = np.sum(sales)
dolya = obshayaviruchka / obshaya_po_magazinu * 100

print("obshaya viruchka magazina: ", obshaya_po_magazinu)
print("dolya regionov v procentah: ", dolya)
lider = np.argmax(obshayaviruchka)
print("krupneyshiy region: ", region_names[lider])
print("dolya krupneyshego regiona: ", dolya[lider], "%")
```

```python
pervye4 = np.mean(sales[:, :4], axis=1)
poslednie4 = np.mean(sales[:, 4:], axis=1)
izmenenie = poslednie4 - pervye4

procent_izmeneniya = np.divide(
    izmenenie * 100,
    pervye4,
    out=np.full(pervye4.shape, np.nan, dtype=float),
    where=pervye4 != 0
)

print("srednie za pervie 4 nedeli: ", pervye4)
print("srednie za poslednie 4 nedeli: ", poslednie4)
print("absolutnoe izmenenie: ", izmenenie)
print("izmenenie v procentah: ", procent_izmeneniya)
```

```python
sales_bez_vibrosov = np.where(vybros, np.nan, sales)
viruchka_bez_vibrosov = np.nansum(sales_bez_vibrosov, axis=1)
srednie_bez_vibrosov = np.nanmean(sales_bez_vibrosov, axis=1)

print("viruchka do udaleniya: ", obshayaviruchka)
print("viruchka bez vibrosov: ", viruchka_bez_vibrosov)
print("srednie do udaleniya: ", srednie)
print("srednie bez vibrosov: ", srednie_bez_vibrosov)
print("reyting do udaleniya: ", region_names[np.argsort(-obshayaviruchka)])
print("reyting bez vibrosov: ", region_names[np.argsort(-viruchka_bez_vibrosov)])
```

```python
est_snizhenie = poslednie4 < pervye4
bolshoy_razbros = koeff_razbrosa > 10
problemniy_region = est_snizhenie & bolshoy_razbros

print("est snizhenie prodaj: ", est_snizhenie)
print("razbros bolshe 10 procentov: ", bolshoy_razbros)
print("regioni dlya proverki: ", region_names[problemniy_region])
```
