```python
import numpy as np 
np.random.seed(42)

num_days = 364
num_products = 4

daily_sales = np.random.randint(low = 10, high = 101, size = (num_days, num_products))

print("Shape:", daily_sales.shape) 
print("First 5 days:\\n", daily_sales[:5])
```

```python
import numpy as np
num_days = 360
num_products = 4


daily_sales = np.random.randint(low=10, high=101, size=(num_days, num_products))

monthly_sales = daily_sales.reshape(12, 30, 4)

monthly_avg_sales = monthly_sales.mean(axis=1)

print("Monthly Avg Sales Shape:", monthly_avg_sales.shape)  # Ожидаемый результат: (12, 4)
print("Monthly Averages:\n", monthly_avg_sales)
```
