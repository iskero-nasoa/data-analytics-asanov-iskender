```python
import numpy as np

order_id = np.array([1001, 1002, 1003, 1004, 1005, 1006])
quantity = np.array([2, 3, 0, 5, -1, 4])
unit_price = np.array([1500, 2000, 1200, 3000, 2500, np.nan])
discount = np.array([1.3, 0.05, 0.00, 0.15, 0.10, 0.20])

print("razmernost: ", order_id.ndim, "forma: ", order_id.shape, "kolichestvo_elementov: ", order_id.size, "tip: ", order_id.dtype)
print("razmernost: ", quantity.ndim, "forma: ", quantity.shape, "kolichestvo_elementov: ", quantity.size, "tip: ", quantity.dtype)
print("razmernost: ", unit_price.ndim, "forma: ", unit_price.shape, "kolichestvo_elementov: ", unit_price.size, "tip: ", unit_price.dtype)
print("razmernost: ", discount.ndim, "forma: ", discount.shape, "kolichestvo_elementov: ", discount.size, "tip: ", discount.dtype)


if order_id.size == quantity.size == unit_price.size == discount.size:
    print("mojno")
else:
    print("nelza")
```

```python
a = order_id[quantity <= 0]
print("nol ili menshe: ", a)
b = order_id[np.isnan(unit_price)]
print("s propushenymi: ", b)
c1 = discount[discount < 0]
c2 = discount[discount > 1]
print("skidki vne dopustimogo diapasona: ", c1, c2)
if a.all() and b.all() and c1.all() and c2.all():
    print("proshol")
```
